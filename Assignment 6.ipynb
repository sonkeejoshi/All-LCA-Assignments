```python
import numpy as np
import pandas as pd

# Create sample data
sample_data = {
    'Name': ['Aman', 'Riya', 'Soham', 'Anam', 'Neha', None, 'Karan', 'Riya'],
    'Marks': [85, 90, None, 85, 78, 92, None, 90],
    'City': [
        'Pune',
        'Mumbai',
        'Delhi',
        'Pune',
        None,
        'Chennai',
        'Delhi',
        'Mumbai',
    ],
}
pd.DataFrame(sample_data).to_csv('students.csv', index=False)

df = pd.read_csv('students.csv')
print('Original DataFrame:')
print(df)
print('\n')
```

```python
# 1. Check basic information
print('Info:')
print(df.info())
print('\n')

print('Description:')
print(df.describe(include='all'))
print('\n')
```

```python
# 2. Handle missing values
# Fill missing numeric values with mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Fill missing categorical values with mode
df['Name'] = df['Name'].fillna(df['Name'].mode()[0])
df['City'] = df['City'].fillna(df['City'].mode()[0])

print('After handling missing values:')
print(df)
print('\n')
```

```python
# 3. Remove duplicate rows
df = df.drop_duplicates()
print('After removing duplicates:')
print(df)
print('\n')
```

```python
# 4. Filter data
selected_data = df[df['Marks'] > 85]
print('Students with marks greater than 85:')
print(selected_data)
print('\n')
```
